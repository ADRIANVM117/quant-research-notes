##  **comparar un straddle con un strangle**

- ¿pagar una prima menor compensa necesitar un movimiento mayor para ganar?


Usa una sola fecha y un mismo vencimiento de tus datos:

- **Straddle:** compra call y put del mismo strike, el más cercano al spot.

- **Strangle:** compra una put con strike inferior al spot y una call con strike superior. Selecciona los strikes disponibles más cercanos a −3% y +3% del spot.


Usa los **ask** para calcular el coste de compra de ambas estructuras.

Ejercicio consiste en:

1. Mostrar los cuatro contratos y comparar las primas totales.
2. Calcular el resultado al vencimiento para movimientos de SPY entre −10% y +10%.
