## Protocolo fijado antes de ejecutar

Se usan exclusivamente filas elegibles comunes. Se calculan correlaciones de Spearman de `future_vol10` con `vol20_t`, A y B. Las diferencias principales son A menos `vol20` y B menos A. La incertidumbre usa bootstrap móvil no circular: bloques de 20 decisiones, 5,000 réplicas y semilla 20260927. Un intervalo que incluye cero es inconcluso respecto a mejora.

```python
from pathlib import Path
import sys, subprocess, json
import pandas as pd
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
REPORTS = ROOT / 'reports'
run = subprocess.run([sys.executable, 'scripts/evaluate_hmm_ab_future_vol10_spearman.py'], cwd=ROOT, text=True, capture_output=True)
print(run.stdout)
if run.stderr: print(run.stderr)
assert run.returncode == 0, run.stderr
```

## Resultados visibles

Los mismos índices de filas se remuestrean para resultado y los tres predictores en cada réplica.

```python
summary = pd.read_csv(REPORTS / 'hmm_ab_future_vol10_spearman_exploratory_summary.csv')
metadata = json.loads((REPORTS / 'hmm_ab_future_vol10_spearman_exploratory_metadata.json').read_text())
display(summary)
print('Filas comunes:', metadata['rows_selected_common'])
print('Fechas:', metadata['decision_date_start'], 'a', metadata['decision_date_end'])
print('Bootstrap:', metadata['replications'], 'réplicas; bloques:', metadata['block_length_decisions'], '; semilla:', metadata['seed'])
```

## Interpretación limitada

Los dos intervalos principales incluyen cero; bajo el protocolo, ambos son inconclusos respecto a mejora. Esto no prueba ni refuta utilidad económica, no transforma los scores en alertas y no constituye una evaluación prospectiva intacta.
