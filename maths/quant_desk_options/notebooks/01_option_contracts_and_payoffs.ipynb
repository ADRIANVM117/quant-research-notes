# Representación de contratos

- Vamos a crear una cadena sintética de opciones sobre un único subyacente.

```python
import numpy as np
import pandas as pd

quote_ts = pd.Timestamp("2026-10-23T14:30:00Z")

rows = [
    ("CALL",  95,  6.10, 6.40),
    ("PUT",   95,  0.90, 1.05),
    ("CALL", 100,  3.10, 3.35),
    ("PUT",  100,  2.90, 3.15),
    ("CALL", 105,  1.25, 1.45),
    ("PUT",  105,  6.20, 6.55),

    # Errores intencionales
    ("CALL", 110,  1.60, 1.40),
    ("PUT",  110, -0.10, 1.20),
]

chain = pd.DataFrame(
    rows,
    columns=[
        "option_type",
        "strike",
        "bid",
        "ask"
    ]
)

chain["underlying"] = "XYZ"
chain["spot"] = 100.0
chain["expiration"] = pd.Timestamp("2026-11-20")
chain["quote_ts"] = quote_ts

chain["style"] = "AMERICAN"
chain["settlement"] = "PHYSICAL"
chain["multiplier"] = 100
chain["currency"] = "USD"
chain["source"] = "SYNTHETIC"

chain
```
